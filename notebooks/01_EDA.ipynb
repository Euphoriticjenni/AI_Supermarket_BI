## Key Findings

1. **Seasonality dominates revenue** — December runs ~20% above the monthly average and February ~10% below; staffing, promotions and inventory should follow this calendar.
2. **Weekends outperform weekdays by ~25%** per day — weekend-specific staffing and stock depth are justified.
3. **Food and beverages is the top revenue line (~25% of revenue)** and also the highest-frequency purchase — it is the anchor category for the assortment.
4. **Profit is almost uniform across branches** — the fixed 4.76% margin means branch ranking is driven entirely by volume, so growth levers are traffic and basket size, not pricing.
5. **Members generate ~65% of transactions** — the loyalty program carries the business; converting the remaining 35% is the largest single upside.
6. **The 5pm peak is the strongest hour** with a secondary lunchtime bump at 1pm — shift schedules and checkout capacity should target 12-13 and 16-18.
7. **Ratings cluster near 7/10 with little branch difference** — service quality is consistent; rating does not explain branch revenue gaps.
